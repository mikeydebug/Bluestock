Validates NAV positivity, enforces non-trading day policies without forward-fill distortion, builds dimensional tables, and populates the SQLite Star Schema.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/db/mutual_funds.db")
print("Connected to SQLite Database.")

tables = ["dim_fund", "dim_date", "fact_nav", "fact_transactions", "fact_aum", "fact_performance"]
for tbl in tables:
    count = pd.read_sql(f"SELECT COUNT(*) as cnt FROM {tbl}", conn)["cnt"][0]
    print(f"Table {tbl:<20}: {count:>8,} records")
```
