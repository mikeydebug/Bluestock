```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

conn = sqlite3.connect("../data/db/mutual_funds.db")
fact_tx = pd.read_sql("SELECT * FROM fact_transactions", conn)
fact_nav = pd.read_sql("SELECT * FROM fact_nav", conn)
dim_fund = pd.read_sql("SELECT * FROM dim_fund", conn)

print(f"Loaded {len(fact_tx):,} transactions and {len(fact_nav):,} NAV records for EDA.")
```

## Key Finding Summary Table: Inflows by State and City Tier

```python
geo_summary = fact_tx.groupby(["city_tier", "investor_state"])["amount"].agg(["count", "sum"]).reset_index()
geo_summary["sum_lakhs"] = geo_summary["sum"] / 1e5
geo_summary.sort_values(by="sum", ascending=False, inplace=True)
print(geo_summary.head(10).to_string(index=False))
```
