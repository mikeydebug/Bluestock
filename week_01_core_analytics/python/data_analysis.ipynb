## 1. Core Python Fundamentals: Variables, Data Structures & Functions
Demonstrating foundational programming building blocks essential for financial data pipelines.

```python
# Python Variables & Types
ticker = "BLUESTOCK-INDEX"
base_nav = 100.0
is_active = True

# Lists & Dictionaries
scheme_codes = [125497, 119551, 120503, 118632, 119092, 120841]
fund_metadata = {
    125497: {"name": "SBI Small Cap Fund", "expense_ratio": 0.68, "aum_cr": 28450},
    118632: {"name": "Nippon India Large Cap Fund", "expense_ratio": 0.82, "aum_cr": 34120},
    120841: {"name": "Quant Mid Cap Fund", "expense_ratio": 0.74, "aum_cr": 12900}
}

# Custom Function with Type Hints & Docstring
def calculate_annualized_return(start_nav: float, end_nav: float, years: float) -> float:
    """Calculates Compound Annual Growth Rate (CAGR)."""
    if start_nav <= 0 or years <= 0:
        raise ValueError("NAV and years must be strictly positive.")
    return (end_nav / start_nav) ** (1.0 / years) - 1.0

# Iteration & Loop Demonstration
print("--- Fund Metadata Summary ---")
for code, meta in fund_metadata.items():
    print(f"Scheme {code}: {meta['name']} | AUM: ₹{meta['aum_cr']:,} Cr | TER: {meta['expense_ratio']}%")

test_cagr = calculate_annualized_return(100.0, 156.25, 3.0)
print(f"
Sample 3-Year CAGR: {test_cagr * 100:.2f}%")
```

## 2. Numerical Computing with NumPy
Vectorized financial arrays, return series simulation, and volatility computation.

```python
import numpy as np

# Simulate 252 daily log returns (1 trading year) with mu=12% annual, sigma=18% annual
np.random.seed(42)
daily_mu = 0.12 / 252
daily_sigma = 0.18 / np.sqrt(252)
daily_returns = np.random.normal(daily_mu, daily_sigma, 252)

# Compute Statistics with NumPy
mean_daily = np.mean(daily_returns)
std_daily = np.std(daily_returns, ddof=1)
ann_volatility = std_daily * np.sqrt(252)
ann_return = (1 + mean_daily) ** 252 - 1

print(f"Mean Daily Return: {mean_daily * 100:.4f}%")
print(f"Annualized Return: {ann_return * 100:.2f}%")
print(f"Annualized Volatility: {ann_volatility * 100:.2f}%")
```

## 3. Data Ingestion, Cleaning & KPIs with Pandas
Loading commercial sales data, evaluating data hygiene, handling nulls/types, and deriving KPIs.

```python
import pandas as pd

# Load Dataset
csv_path = "../eda/cleaned_dataset.csv"
df = pd.read_csv(csv_path)
df["order_date"] = pd.to_datetime(df["order_date"])

print(f"Dataset Shape: {df.shape}")
print("
--- Data Types & Non-Null Counts ---")
print(df.info())

# Financial KPIs
total_gross = df["gross_revenue"].sum()
total_net = df["net_revenue"].sum()
total_profit = df["net_profit"].sum()
overall_margin = (total_profit / total_net) * 100
avg_order_val = df["net_revenue"].mean()

print("
================ FINANCIAL KPIS ================")
print(f"Total Gross Sales:     INR {total_gross:,.2f}")
print(f"Total Net Revenue:     INR {total_net:,.2f}")
print(f"Total Net Profit:      INR {total_profit:,.2f}")
print(f"Overall Profit Margin: {overall_margin:.2f}%")
print(f"Average Order Value:   INR {avg_order_val:,.2f}")
print("================================================")
```

## 4. Multi-Dimensional Business Aggregations
Segment and regional analysis using Pandas  and pivot tables.

```python
# Group by Segment
segment_kpis = df.groupby("customer_segment").agg(
    Orders=("order_id", "count"),
    Net_Revenue=("net_revenue", "sum"),
    Avg_Order_Value=("net_revenue", "mean"),
    Net_Profit=("net_profit", "sum")
).sort_values(by="Net_Revenue", ascending=False)

segment_kpis["Margin_Pct"] = (segment_kpis["Net_Profit"] / segment_kpis["Net_Revenue"]) * 100
print("--- Performance by Customer Segment ---")
print(segment_kpis.to_string())
```

## 5. Visualizations with Matplotlib
Plotting revenue distributions, regional performance, and profit margins.

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Chart 1: Revenue by Region
region_rev = df.groupby("region")["net_revenue"].sum() / 1e5
axes[0].bar(region_rev.index, region_rev.values, color=["#1f77b4", "#2ca02c", "#ff7f0e", "#d62728"])
axes[0].set_title("Net Revenue by Geographic Region (₹ Lakhs)", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Region")
axes[0].set_ylabel("Net Revenue (₹ Lakhs)")
axes[0].grid(axis="y", linestyle="--", alpha=0.7)

# Chart 2: Product Net Profit Distribution
prod_profit = df.groupby("product_name")["net_profit"].sum() / 1e5
prod_profit.sort_values().plot(kind="barh", ax=axes[1], color="#2ca02c")
axes[1].set_title("Net Profit Contribution by Product (₹ Lakhs)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Net Profit (₹ Lakhs)")
axes[1].grid(axis="x", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```
